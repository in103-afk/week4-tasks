```python
#import and install 
!pip3 install textblob
!python3 -m textblob.download_corpora
from textblob import TextBlob 

#creating the first textblob 
wiki = TextBlob("Python is a high-level, general-purpose programming language.")
```

## Part -of -speech tagging

```python
wiki.tags
```

## Noun Phrase Extraction

```python
wiki.noun_phrases
```

## Sentiment Analysis

```python
#gives polarity and subjectivity scores 
testimonial = TextBlob("Textblob is amazingly simple to use. What great fun!") 
testimonial.sentiment 
testimonial.sentiment.polarity
```

## Tokenization

```python
zen = TextBlob(
    "Beautiful is better than ugly. "
    "Explicit is better than implicit. "
    "Simple is better than complex."
)

zen.words 
zen.sentences 
for sentence in zen.sentences:
    print(sentence.sentiment)

```

## Words, inflection, and lemmatization 

```python
sentence = TextBlob("Use 4 spaces per indentation level.") 
sentence.words 
sentence.words[2].singularize()
sentence.words[-1].pluralize() 
from textblob import Word
w = Word("octopi")
w.lemmatize() 
w = Word("went")
w.lemmatize("v") 
```

## WordNet Integration

```python
from textblob import Word
from textblob.wordnet import VERB
word = Word("octopus")
word.synsets 
Word("hack").get_synsets(pos=VERB)
#can access the definitions for each synset by the define property or the define() 
Word("octopus").definitions 
#can also get synsets directly 
from textblob.wordnet import Synset 
octopus = Synset("octopus.n.02") 
shrimp = Synset("shrimp.n.03") 
octopus.path_similarity(shrimp)
```

## WordLists 

```python
#wordlist is a python list with additional methods 
animals = TextBlob("cat dog octopus") 
animals.words 
animals.words 

```

## Spelling Correction

```python
b = TextBlob("I havv goood speling!")  
print(b.correct()) 
#word objects have a spellcheck method that returns a list of tuples with spelling suggestions 
from textblob import Word 
w = Word("falibility") 
w.spellcheck() 
#spell check is 90 % correct tutorial says   
```

## Get Word and Noun Phrase Frequencies

```python
monty = TextBlob("We are no longer the Knights who say Ni. "
                    "We are now the Knights who say Ekki ekki ekki PTANG.")
monty.word_counts['ekki'] 

#count method to get frequency of words and phrases 
monty.word_counts['ekki']  

#can specify whether or not search needs to be case sensitive 
monty.words.count('ekki', case_sensitive=True) 
wiki.noun_phrases.count('python')
```

## Parsing 

```python
b = TextBlob("And now for something completely different.") 
print(b.parse()) 
#pattern's parses 
```

## TextBlobs Are Like Python Strings!

```python
zen[0:19] 
#can use common string methods 
zen.upper()
zen.find("Simple") 

#make comparisons between textblobs and strings 
apple_blob = TextBlob("apples") 
banana_blob = TextBlob("bananas") 
apple_blob < banana_blob 
apple_blob == "apples" 

#you can concatonate and interpolate TextBlobs and strings 
apple_blob + " and " + banana_blob 
"{0} and {1}".format(apple_blob, banana_blob)
```

## n-grams 

```python
#this returns a tuple of n successive words 
blob = TextBlob("Now is better than never.") 
blob.ngrams(n=3) 
```

## Get Start and End Indices of Sentences 

```python
#use sentence.start and sentence.end to get start and end indices of sentences 
for s in zen.sentences:
    print(s)
    print("---- Starts at index {}, Ends at index {}".format(s.start, s.end))
```
